## 1. PRÉSENTATION DU PROJET

### 1.1 Contexte

Les établissements scolaires privés en Afrique francophone (écoles primaires, collèges, lycées) rencontrent des difficultés majeures dans la gestion administrative et financière de leurs élèves. La gestion manuelle (cahiers, Excel éparpillés) génère :

- **Perte de temps** : 45 minutes par inscription
- **Erreurs de facturation** : Confusion dans les paiements multi-enfants
- **Impayés élevés** : Absence de suivi rigoureux
- **Dispersion des données** : Informations académiques et financières non centralisées
- **Difficulté de reporting** : Impossibilité d'avoir une vision globale

### 1.2 Objectifs du projet

Développer une **application desktop (WPF) .NET 8** permettant aux établissements de :

1. **Centraliser** toutes les données scolaires et financières
2. **Automatiser** le suivi des paiements par famille
3. **Accélérer** les processus d'inscription (de 45 min à 8 min)
4. **Améliorer** le taux de recouvrement (+25% visé)
5. **Générer** automatiquement les documents officiels

### 1.3 Périmètre

**Inclus dans le MVP :**
- Gestion des familles et élèves
- Suivi financier complet (frais, paiements, impayés)
- Gestion académique basique (notes, emploi du temps)
- Génération de documents (carte élève, reçus, bulletins)
- Système de rôles et permissions
- Compatibilité Windows 7 à 11

**Exclu du MVP (Phases ultérieures) :**
- Portail parents en ligne
- Application mobile
- Intégration SMS/Email automatisée
- Module LMS (e-learning)
- Gestion RH complète

## 3. SOLUTION PROPOSÉE

### 3.1 Description générale

**GMS (Gestion Management School)** est une application desktop Windows qui centralise toute la gestion administrative et financière d'un établissement scolaire, avec un focus particulier sur le suivi familial multi-enfants.

### 3.2 Architecture technique

```
┌─────────────────────────────────────────────────┐
│           APPLICATION WPF (.NET 8)              │
├─────────────────────────────────────────────────┤
│  MVVM Pattern + CQRS (MediatR)                  │
├─────────────────────────────────────────────────┤
│  Application Layer (Logique métier)             │
├─────────────────────────────────────────────────┤
│  Infrastructure Layer (EF Core, Repositories)   │
├─────────────────────────────────────────────────┤
│  Domain Layer (Entités métier)                  │
├─────────────────────────────────────────────────┤
│          SQL SERVER EXPRESS (Local)             │
└─────────────────────────────────────────────────┘
```

### 3.3 Stack technologique

| Couche | Technologie | Justification |
|--------|-------------|---------------|
| **Frontend** | WPF (.NET 8) | Interface riche, compatible Win 7-11 |
| **Backend** | C# .NET 8 | Performance, maturité, LTS |
| **Base de données** | SQL Server Express | Gratuit, robuste, hors-ligne |
| **ORM** | Entity Framework Core 8 | Productivité, migrations |
| **Architecture** | CQRS (MediatR) | Séparation lecture/écriture |
| **Validation** | FluentValidation | Règles métier strictes |
| **Sécurité** | BCrypt.Net | Hash passwords sécurisés |
| **PDF** | QuestPDF | Génération documents |
| **Conteneurisation** | Docker | Déploiement simplifié |
| **CI/CD** | GitHub Actions | Automatisation builds |

## 4. BESOINS FONCTIONNELS

### 4.1 Gestion des utilisateurs (BF-01)

**Description :** Système d'authentification avec 4 rôles distincts

**Acteurs :** Tous les utilisateurs

**Fonctionnalités :**
- Connexion sécurisée (username/password)
- 4 rôles : Admin, Directeur, Secrétaire, Comptable
- Permissions granulaires par rôle
- Changement de mot de passe
- Journalisation des connexions

**Règles métier :**
- Mot de passe minimum 8 caractères (1 majuscule, 1 chiffre)
- Verrouillage après 5 tentatives échouées
- Session expire après 8h d'inactivité

**Priorité :** P0 (Critique)

---

### 4.2 Gestion des familles (BF-02)

**Description :** Centralisation des informations familiales

**Acteurs :** Secrétaire, Comptable, Directeur

**Fonctionnalités :**
- Création fiche famille (père, mère, contacts, adresse)
- Vision consolidée : Total dû, Total payé, Solde global
- Liste de tous les enfants de la famille
- Historique complet des paiements familiaux
- Recherche par nom/téléphone

**Règles métier :**
- Minimum 1 parent renseigné
- Téléphone obligatoire pour relances
- Impossibilité de supprimer une famille avec élèves actifs

**Priorité :** P0 (Critique)

---

### 4.3 Gestion des élèves (BF-03)

**Description :** Dossier scolaire complet par élève

**Acteurs :** Secrétaire, Directeur

**Fonctionnalités :**

**Inscription :**
- Formulaire complet (nom, prénom, date naissance, sexe, etc.)
- Rattachement à une famille existante ou nouvelle
- Upload photo élève
- Affectation à une classe
- Génération automatique du matricule unique (EL202400123)

**Dossier :**
- Onglet Informations générales
- Onglet Finances (frais, paiements, solde)
- Onglet Notes académiques
- Onglet Emploi du temps
- Onglet Documents (carte, certificat)

**Règles métier :**
- Matricule auto-généré et unique
- Âge minimum 3 ans, maximum 25 ans
- Photo optionnelle mais recommandée
- Génération automatique des frais standards à l'inscription

**Priorité :** P0 (Critique)

---

### 4.4 Gestion financière (BF-04)

**Description :** Suivi complet des finances par élève et famille

**Acteurs :** Comptable, Secrétaire, Directeur

**Fonctionnalités :**

**Configuration des frais :**
- Création types de frais (inscription, scolarité, cantine, etc.)
- Montants par niveau/classe
- Frais récurrents vs ponctuels
- Échéancier personnalisable

**Facturation :**
- Génération automatique des frais à l'inscription
- Ajout manuel de frais supplémentaires
- Bulletin financier par élève
- Vue par échéance

**Paiements :**
- Enregistrement paiement (date, montant, mode)
- Ventilation automatique entre enfants d'une famille
- Génération reçu imprimable
- Historique complet

**Tableau de bord :**
- Total à encaisser
- Total encaissé
- Taux de recouvrement
- Liste des impayés par famille
- Top 10 des plus gros impayés

**Règles métier :**
- Paiement ne peut être daté dans le futur
- Ventilation totale = montant payé (contrôle strict)
- Impossibilité de supprimer un paiement (soft delete)
- Recalcul automatique des soldes

**Priorité :** P0 (Critique)

---

### 4.5 Gestion académique (BF-05)

**Description :** Suivi des notes et emplois du temps

**Acteurs :** Secrétaire, Directeur

**Fonctionnalités :**

**Notes :**
- Saisie notes par matière et période (trimestre)

**Emploi du temps :**
- Création emploi du temps par classe
- Affichage par jour/semaine

**Règles métier :**
- Notes entre 0 et 20 (ou barème personnalisé)

**Priorité :** P1 (Important)

## 7. MODÈLE DE DONNÉES (Simplifié)

### 7.1 Entités principales

```
FAMILLE (1) ──< (N) ELEVE (N) >── (1) CLASSE
    │                │
    │                │
    V                V
PAIEMENT        FRAIS (N) >── (1) TYPE_FRAIS
    │                │
    │                V
    └────>  VENTILATION_PAIEMENT

ELEVE (1) ──< (N) NOTE (N) >── (1) MATIERE
    │                             │
    V                             V
PERIODE                    EMPLOI_DU_TEMPS
```

### 7.2 Cardinalités clés

- **1 Famille** → **N Élèves** (famille nombreuse)
- **1 Élève** → **N Frais** (plusieurs types de frais)
- **1 Paiement** → **N Ventilations** (répartition multi-enfants)
- **1 Élève** → **1 Classe** (à un instant T)
- **1 Classe** → **N Élèves** (effectif classe)

## 8. ARCHITECTURE LOGICIELLE

### 8.1 Pattern architectural : Clean Architecture + CQRS

```
┌────────────────────────────────────────────┐
│  Presentation Layer (WPF)                  │
│  - Views (XAML)                            │
│  - ViewModels (MVVM)                       │
│  - Commands/Bindings                       │
└────────────────┬───────────────────────────┘
                 │
┌────────────────▼───────────────────────────┐
│  Application Layer                         │
│  - CQRS (Commands/Queries)                 │
│  - Handlers (MediatR)                      │
│  - DTOs                                    │
│  - Validators (FluentValidation)           │
└────────────────┬───────────────────────────┘
                 │
┌────────────────▼───────────────────────────┐
│  Domain Layer                              │
│  - Entities                                │
│  - Value Objects                           │
│  - Domain Events                           │
│  - Business Rules                          │
└────────────────┬───────────────────────────┘
                 │
┌────────────────▼───────────────────────────┐
│  Infrastructure Layer                      │
│  - DbContext (EF Core)                     │
│  - Repositories                            │
│  - External Services                       │
└────────────────┬───────────────────────────┘
                 │
┌────────────────▼───────────────────────────┐
│  Database (SQL Server Express)             │
└────────────────────────────────────────────┘
```

### 8.2 Patterns utilisés

| Pattern | Utilité | Implémentation |
|---------|---------|----------------|
| **MVVM** | Séparation UI/Logique | WPF natif |
| **CQRS** | Séparation lecture/écriture | MediatR |
| **Repository** | Abstraction accès données | Generic Repository |
| **Unit of Work** | Transaction atomique | DbContext |
| **Dependency Injection** | Couplage faible | Microsoft.Extensions.DI |
| **Factory** | Création objets complexes | Document generators |
| **Strategy** | Calculs dynamiques | Barèmes notes |
