## RISQUES & PLAN DE CONTINGENCE

### Risque 1 : Retard développement

**Probabilité :** Moyenne  
**Impact :** Élevé

**Plan B :**
- Réduire périmètre MVP (retirer module notes)
- Prolonger sprints de 1 semaine
- Recruter dev freelance temporaire

### Risque 2 : Bugs critiques non détectés

**Probabilité :** Moyenne  
**Impact :** Critique

**Plan B :**
- Phase beta testing prolongée (3 semaines)
- Hotfix release rapide (< 24h)
- Support téléphonique renforcé

### Risque 3 : Adoption lente

**Probabilité :** Élevée  
**Impact :** Élevé

**Plan B :**
- Offre "3 mois gratuits" premiers clients
- Programme parrainage (réduction 20%)
- Démos sur site gratuites
